#### 3. Call the OpenAI API with different `temperature` values

Prompt: "Write a two sentence story about a robot learning to cook."

`temperature` values: `0`, `0.7` and `1.5` three times for each setting.

Compare the outputs.

Tip: With Chat Completions API you can use parameter `n=3` to generate three different responses to your prompt with one call.

````python
def get_responses(model_name, prompt, other_params={}):
    return client.chat.completions.create(
        model=model_name,
        messages=[
            {"role": "user", "content": prompt}
        ],
        **other_params
    ).choices
````

````python
prompt = "Write a two sentence story about a robot learning to cook."
temperatures = [0, 0.7, 1.5]
temp_responses = {}
for i in temperatures:
    temp_responses[i] = get_responses(
        "gpt-4.1-mini",
        prompt,
        {
            "temperature": i,
            "n": 3
        }
    )
````

````python

for temp, responses in temp_responses.items():
    print(f"Temperature = {temp} ")
    for i, choice in enumerate(responses):
        print(f"\nResponse {i+1}:")
        print(choice.message.content)
    print("\n" + "="*100 + "\n")
    
````

### Observations:

**Temperature = 0:** Responses are highly consistent and deterministic. The model tends to generate very similar or identical responses across the three calls, showing predictable behavior. second and third responses are almost similar.

**Temperature = 0.7:** Responses show variability with some consistency comparing to temp = 0. This balanced setting produces diverse but also coherent stories, making it suitable for moderately creative tasks while maintaining quality.

**Temperature = 1.5:** Responses are highly diverse and creative, producing unusual word choices and unexpected plot directions. More prone to producing creative but occasionally less coherent content.

#### 10. Write a Few-Shot prompt to request a sentiment analysis of a product review.

Review: "The product is amazing, and I love using it every day!"

Observe how the model's performance changes and compare the response to Zero-Shot output.

````python
review = "The product is amazing, and I love using it every day!"

few_shot_prompt = f"""Analyze the sentiment of the product reviews below and classify each as positive, negative, or neutral.

Example 1:
Review: "This product is terrible and doesn't work at all."
Sentiment: Negative

Example 2:
Review: "The product is okay, nothing special."
Sentiment: Neutral

Example 3:
Review: "I absolutely love this product! It changed my life."
Sentiment: Positive

Now analyze this review:
Review: "{review}"
Sentiment:"""

response_fewshot = get_response("gpt-4.1-mini", few_shot_prompt)

print("FEW-SHOT SENTIMENT ANALYSIS")
print(f"Review: {review}")
print(f"\nResult: {response_fewshot}")
````

### Comparison: Zero-Shot vs Few-Shot

**Zero-Shot Performance:**
- Relies solely on the model's built-in understanding of sentiment analysis
- Provides a single classification without detailed reasoning
- Works reasonably well for clear cases like this positive review

**Few-Shot Performance:**
- Provides specific examples to guide the model's understanding
- Establishes a clear pattern of how to classify different sentiments
- Model sees examples of positive, negative, and neutral sentiments
- Generally produces more accurate and reliable results

**Conclusion:**<br>
Few-shot prompting typically performs better than zero-shot prompting because it provides the model with concrete examples of the expected output format and classification patterns. The model learns from the examples and can apply that pattern more reliably to new inputs.

##### Response
GENERATED CODE:
```python
def factorial(n: int) -> int:
    """
    Calculate the factorial of a non-negative integer n.
    
    The factorial of n (denoted as n!) is the product of all positive integers 
    less than or equal to n. By definition, 0! is 1.

    Parameters:
    n (int): A non-negative integer whose factorial is to be computed.

    Returns:
    int: The factorial of n.

    Raises:
    ValueError: If n is negative.
    """
    if n < 0:
        raise ValueError("Factorial is not defined for negative integers")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
```
