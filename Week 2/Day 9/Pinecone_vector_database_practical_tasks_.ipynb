```python
from dotenv import load_dotenv
import os
load_dotenv()

# initialize client
pc = Pinecone(api_key=os.getenv("PINECONE_API"))

index_name = "accenture-bootcamp-py"

if not pc.has_index(index_name):
    pc.create_index_for_model(
        name=index_name,
        cloud="aws",
        region="us-east-1",
        embed={
            "model":"llama-text-embed-v2",
            "field_map":{"text": "chunk_text"}
        }
    )
```

```python
index = pc.Index(index_name)

# Upsert text data
text_data = [
    {"_id": "rec1", "chunk_text": "Machine learning is a subset of artificial intelligence that enables systems to learn and improve from experience."},
    {"_id": "rec2", "chunk_text": "Deep learning uses neural networks with multiple layers to process data and extract features."},
    {"_id": "rec3", "chunk_text": "Natural language processing is a field of AI focused on understanding and generating human language."},
    {"_id": "rec4", "chunk_text": "Vector databases store data as high-dimensional vectors, enabling efficient similarity search."},
    {"_id": "rec5", "chunk_text": "Semantic search finds documents based on meaning rather than keyword matching."}
]

# Upsert the data into the index
index.upsert_records("__default__",text_data)
print(f"Successfully upserted {len(text_data)} text documents to the index.")
```

Print search results.

```python
# search
query = "How do neural networks work in deep learning?"

results = index.search(
    "__default__",
    query={
        "top_k": 3,
        "inputs": {
            "text": query,
        }
    }
)

print(f"Query: '{query}'\n")
print("| id    | score | text                                                                    |")
print("|-------|-------|","-"*73,"|")
for hit in results['result']['hits']:
    print(f"| {hit['_id']:<5} | {round(hit['_score'], 2):<5} | {hit['fields']['chunk_text']:<50} |")
```
