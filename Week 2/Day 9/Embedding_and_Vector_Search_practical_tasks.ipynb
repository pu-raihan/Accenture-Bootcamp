Write and test a function named 'cos_similarity_calculation' that determines the semantic similarity between the sentences in your list and any text sentence using their vector representations and the cosine distance as a similarity measure.  

```python
# place your code here
def cos_similarity_calculation(query_sentence, sentences_list, embeddings_list):
    query_embedding = model.encode(query_sentence)
    
    similarities = util.cos_sim(query_embedding, embeddings_list)
    
    # sentences and their similarity scores
    results = []
    for idx, similarity_score in enumerate(similarities[0]):
        results.append((sentences_list[idx], float(similarity_score)))
    
    results.sort(key=lambda x: x[1], reverse=True)
    
    return results

query = "Deep learning and neural networks are the core of artificial intelligence"
results = cos_similarity_calculation(query, sentences_list, embeddings)

print(f"Query: {query}\n")
print("Top 5 most similar sentences:")
for i, (sentence, score) in enumerate(results[:5], 1):
    print(f"{i}. Score: {score:.4f} | {sentence}")
```

Create a function that determines the cosine similarity between a vector and a batch of vectors using the cosine distance formula and the numpy library. Add code to demonstrate how to use this function.

Cosine similarity formula: similarity = (A · B) / (||A|| * ||B||)

Parameters:
vector (array): Single vector of shape (n_features,)
batch_vectors (array): Batch of vectors of shape (n_samples, n_features)
    

```python
# place your code here
import numpy as np
from numpy.linalg import norm

def cosine_similarity_batch(vector, batch_vectors):
    vector_norm = vector / norm(vector)
    
    batch_norm = batch_vectors / norm(batch_vectors, axis=1, keepdims=True)
    
    similarities = np.dot(batch_norm, vector_norm)
    
    return similarities

test_sentences = [
    "Everyone is running behind the healthy lifestyle trend, but I will never give up my love for food and sleep",
    "Artificial intelligence is transforming industries with automated decision making and predictions",
]
test_vectors = model.encode(test_sentences)
similarities1 = cosine_similarity_batch(test_vectors[0], embeddings)
similarities2 = cosine_similarity_batch(test_vectors[1], embeddings)
print("First sentence:\n", test_sentences[0])
print("\nCosine similarity between first test sentence and all sentences:")
for idx, score in enumerate(similarities1):
    print(f"{idx+1}. {sentences_list[idx][:60]}... | Score: {score:.4f}")

top_indices = np.argsort(similarities1)[::-1][:3]
print("\nTop 3 most similar sentences:")
for i, idx in enumerate(top_indices, 1):
    print(f"{i}. Score: {similarities1[idx]:.4f} | {sentences_list[idx]}")
print("\n" + "-"*80 + "\n")
print("\nSecond sentence:\n", test_sentences[1])
print("\nCosine similarity between second test sentence and all sentences:")
for idx, score in enumerate(similarities2):
    print(f"{idx+1}. {sentences_list[idx][:60]}... | Score: {score:.4f}")

top_indices2 = np.argsort(similarities2)[::-1][:3]
print("\nTop 3 most similar sentences for second test sentence:")
for i, idx in enumerate(top_indices2, 1):
    print(f"{i}. Score: {similarities2[idx]:.4f} | {sentences_list[idx]}")
```
